```python
# Test pytrends installation
print("Testing pytrends installation...\n")

try:
    from pytrends.request import TrendReq
    print("✅ Import successful: pytrends.request.TrendReq")
except ImportError as e:
    print(f"❌ Import failed: {e}")
    exit()

# Initialize client
try:
    gt = TrendReq(hl='en-US', tz=360)
    print("✅ TrendReq client initialized")
except Exception as e:
    print(f"❌ Client initialization failed: {e}")
    exit()

# Test: Query a simple trend (GTA VI)
try:
    print("\n🔍 Querying Google Trends for 'GTA VI'...")
    gt.build_payload(['GTA VI'], timeframe='today 1-m')
    data = gt.interest_over_time()
    print("✅ Query successful!")
    print(f"\nData shape: {data.shape}")
    print(f"Columns: {data.columns.tolist()}")
    print(f"\nFirst 5 rows:")
    print(data.head())
except Exception as e:
    print(f"❌ Query failed: {e}")
    exit()

print("\n" + "="*80)
print("✅ PYTRENDS IS WORKING CORRECTLY!")
print("="*80)
```

```python
import pandas as pd

df = pd.read_csv("data/processed/prices_with_returns.csv", parse_dates=["date"])

wide = df.pivot(index="date", columns="ticker", values="return")

# 1) Are returns literally identical?
same_share = (wide["EA"].round(12) == wide["ATVI"].round(12)).mean()
print("Share of days EA == ATVI:", same_share)

# 2) Correlation (should be high-ish but NOT 1.0)
print(wide[["EA", "ATVI"]].corr())

# 3) Quick sanity: first rows
print(wide[["EA", "ATVI"]].head(10))
```
