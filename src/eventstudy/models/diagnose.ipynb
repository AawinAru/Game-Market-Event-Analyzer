```python
import pandas as pd
import matplotlib.pyplot as plt

# === LOAD FILES ===
events = pd.read_csv("regression_dataset_with_gta_trends.csv", sep=';')
trends = pd.read_csv("gta_trends.csv")

print("Events shape:", events.shape)
print("Trends shape:", trends.shape)
print("\nEvent sample:")
print(events.head())

print("\nTrend sample:")
print(trends.head())

# === CHECK MERGE BEHAVIOR ===
merged = events.merge(trends, on="date", how="left", suffixes=("", "_check"))

print("\nColumns after merge:", merged.columns)

# === CHECK GTA trends incorrectly assigned to other publishers ===
non_ttwo = merged[merged["ticker"] != "TTWO"]

print("\nNon-TTWO with GTA trends (should be removed later):")
print(non_ttwo[["ticker","trend_gta","trend_gta6","trend_gta_mean"]].head())

# === DIAGNOSTIC PLOT ===
plt.figure(figsize=(12,5))
plt.plot(trends["date"], trends["trend_gta_mean"], label="GTA Mean Trend")
plt.title("GTA Search Intensity (Raw)")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()

plt.figure(figsize=(12,5))
plt.plot(trends["date"], (trends["trend_gta_mean"]-trends["trend_gta_mean"].mean())/trends["trend_gta_mean"].std(), 
         label="GTA Trend (Z-score)")
plt.title("GTA Trend (Standardized)")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```
